# MovieMind AI — Movie Recommendation System

This notebook demonstrates content-based, collaborative, and hybrid movie recommendations.

```python
import pandas as pd
from recommender import load_data, build_content_model, hybrid_recommendations

movies, ratings = load_data()
movies.head()
```

## 1. Explore the data


```python
print('Movies:', len(movies))
print('Ratings:', len(ratings))
print('Users:', ratings.userId.nunique())
print('Average rating:', round(ratings.rating.mean(), 2))
```

## 2. Content-based filtering
Genres are transformed with TF-IDF and compared using cosine similarity.

```python
_, similarity = build_content_model(movies)
favorite = 'Inception'
movie_id = int(movies.loc[movies.title == favorite, 'movieId'].iloc[0])
from recommender import content_recommendations
content_recommendations(movie_id, movies, similarity, 5)[['title','genres','score']]
```

## 3. Hybrid recommendation
The final score combines content similarity and collaborative preference information.

```python
result = hybrid_recommendations(
    user_id=1,
    favorite_movie_id=movie_id,
    movies=movies,
    ratings=ratings,
    content_similarity=similarity,
    n=5,
    content_weight=0.6
)
result[['title','genres','content_score','collab_score','hybrid_score']]
```
